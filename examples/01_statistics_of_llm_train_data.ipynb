## 1. Number of Tokens

```python
import datasets
from textanalyzer.processors import TokenCountProcessor
```

* load datasets

```python
ds = datasets.load_dataset('Open-Orca/slimorca-deduped-cleaned-corrected', split='train')
```

```python
text_ds = ds.map(lambda example: {'text': ' '.join([conv['value'] for conv in example['conversations']])})
```

* Defualt `TokenCountProcessor` is tiktoken's `o200k_base` tokenizer.

```python
tc_proc = TokenCountProcessor()
text_count_ds = text_ds.map(tc_proc, batched=True, fn_kwargs={'text_column': 'text'})
```

* get token statistics

```python
token_stats = tc_proc.compute_stats(text_count_ds['token_count'])
token_stats
```

```python
print(f"Number of tokens: {tc_proc.format_token_counts(token_stats['total_tokens'])}")
```

You can use another tokenizer to count the tokens.

Please refer the `TokenFactory`.
